```python
import torch
import math
dtype = torch.float
device = torch.device("mps")

# Create random input and output data
x = torch.linspace(-math.pi, math.pi, 2000, device=device, dtype=dtype)
y = torch.sin(x)

# Randomly initialize weights
a = torch.randn((), device=device, dtype=dtype)
b = torch.randn((), device=device, dtype=dtype)
c = torch.randn((), device=device, dtype=dtype)
d = torch.randn((), device=device, dtype=dtype)

learning_rate = 1e-6
for t in range(5000):
    # Forward pass: compute predicted y
    y_pred = a + b * x + c * x ** 2 + d * x ** 3

    # Compute and print loss
    loss = (y_pred - y).pow(2).sum().item()
    if t % 100 == 99:
        print(t, loss)

# Backprop to compute gradients of a, b, c, d with respect to loss
    grad_y_pred = 2.0 * (y_pred - y)
    grad_a = grad_y_pred.sum()
    grad_b = (grad_y_pred * x).sum()
    grad_c = (grad_y_pred * x ** 2).sum()
    grad_d = (grad_y_pred * x ** 3).sum()

    # Update weights using gradient descent
    a -= learning_rate * grad_a
    b -= learning_rate * grad_b
    c -= learning_rate * grad_c
    d -= learning_rate * grad_d


print(f'Result: y = {a.item()} + {b.item()} x + {c.item()} x^2 + {d.item()} x^3')
```

```python
import torch
from torch.utils.data import Dataset
import xarray as xr
import torch.nn.functional as F

class OceanDataset(Dataset):
    def __init__(self,
                 so_files, thetao_files, uo_files, vo_files, zos_files,
                 tas_files, uas_files, vas_files, huss_files, rsns_files):

        self.so = so_files
        self.thetao = thetao_files
        self.uo = uo_files
        self.vo = vo_files
        self.zos = zos_files

        self.tas = tas_files
        self.uas = uas_files
        self.vas = vas_files
        self.huss = huss_files
        self.rsns = rsns_files

    def __len__(self):
        return 1 ## 기존 return len(self.so) - 4
    
    def __getitem__(self, idx):
        
        # ocean
        ds_o = xr.open_dataset(self.so[0])
        so = ds_o['so'].isel(time=0, lev=0).values

        # atm
        ds_a = xr.open_dataset(self.tas[0])
        tas = ds_a['tas'].isel(time=0).values

        # tensor
        so = torch.tensor(so).unsqueeze(0)
        tas = torch.tensor(tas).unsqueeze(0)

        # 사이즈 맞추기
        so = F.interpolate(so.unsqueeze(0), size=(180, 360)).squeeze(0)
        tas = F.interpolate(tas.unsqueeze(0), size=(180, 360)).squeeze(0)

        x = torch.cat([so, tas], dim=0)  # (C, H, W)

        x = x.unsqueeze(0)  # (T=1, C, H, W)

        y = tas  # (1, H, W) 또는 so

        return x, y

    # def __getitem__(self, idx):
    #     x_seq = []

    #     for i in range(4):  # t-3 ~ t

    #         # ===== ocean =====
    #         so = xr.open_dataset(self.so[idx+i])['so'].isel(time=0, lev=0).values
    #         th = xr.open_dataset(self.thetao[idx+i])['thetao'].isel(time=0, lev=0).values
    #         uo = xr.open_dataset(self.uo[idx+i])['uo'].isel(time=0, lev=0).values
    #         vo = xr.open_dataset(self.vo[idx+i])['vo'].isel(time=0, lev=0).values
    #         zos = xr.open_dataset(self.zos[idx+i])['zos'].isel(time=0).values

    #         # ===== atm =====
    #         tas = xr.open_dataset(self.tas[idx+i])['tas'].isel(time=0).values
    #         uas = xr.open_dataset(self.uas[idx+i])['uas'].isel(time=0).values
    #         vas = xr.open_dataset(self.vas[idx+i])['vas'].isel(time=0).values
    #         huss = xr.open_dataset(self.huss[idx+i])['huss'].isel(time=0).values
    #         rsns = xr.open_dataset(self.rsns[idx+i])['rsns'].isel(time=0).values

    #         # ===== tensor =====
    #         ocean = torch.tensor([th, so, uo, vo, zos], dtype=torch.float32)
    #         atm   = torch.tensor([tas, uas, vas, huss, rsns], dtype=torch.float32)

    #         x = torch.cat([ocean, atm], dim=0)  # (C, H, W)

    #         x_seq.append(x)

    #     x = torch.stack(x_seq)  # (T, C, H, W)

    #     # ===== downsampling =====
    #     x = F.interpolate(x, size=(180, 360))

    #     # ===== target (zos 예측) =====
    #     y = xr.open_dataset(self.zos[idx+4])['zos'].isel(time=0).values
    #     y = torch.tensor(y, dtype=torch.float32).unsqueeze(0)
    #     y = F.interpolate(y.unsqueeze(0), size=(180, 360)).squeeze(0)

    #     return x, y
```
